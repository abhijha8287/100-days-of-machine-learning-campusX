```python
import numpy as np 
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer
%matplotlib inline
```

```python
tips=sns.load_dataset('tips')  
```

```python
titanic=pd.read_csv('titanic/train.csv')
```

```python
titanic.head()
```

```python
flights=sns.load_dataset('flights')
```

```python
iris=sns.load_dataset('iris')
```

```python
# scatter plot
# numerical - numerical
sns.scatterplot(x='total_bill',y='tip',data=tips, hue='sex', style='smoker',size='size')
```

```python
#bar plot
#Numerical -Categorical
sns.barplot(x='Pclass', y='Age', data=titanic, hue='Sex')
```

```python
# box plot
# numerical -caterogical
sns.boxplot(x='Sex', y='Age', data=titanic, hue='Survived')
```

```python
# dist plot
# numerical - categorical
sns.distplot(titanic[titanic['Survived']==0]['Age'], hist=False)
sns.distplot(titanic[titanic['Survived']==1]['Age'], hist=False)
```

```python
#heatmap
#Categorical-Categorical
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
(titanic.groupby('Pclass')['Survived'].mean() * 100).plot(kind='bar')
```

```python
# clustermap
# catergorical-categorical
sns.clustermap(pd.crosstab(titanic['Parch'],titanic['Survived']))
```

```python
# pairplot
sns.pairplot(iris, hue='species')
```

```python
# Line plot
sns.lineplot(x='year', y='passengers', data=flights, hue='month')
```
